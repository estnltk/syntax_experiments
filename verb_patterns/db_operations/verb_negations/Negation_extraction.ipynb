### Function extract_negations()

Function `extract_negations()` extracts ID-s of transactions that contain a verb negation based on given negation patterns. The result is a new table of IDs of transaction matches and corresponding pattern IDs.

Required parameters are:

    cur - SQLite Cursor object,
    transaction_head - transaction head table name,
    transaction_row - transaction row table name,
    patterns - negation patterns table name,
    output_table - output table name
    
Voluntary parameters are:

    overwrite - whether output table should be overwritten, if it exists (default False)
    
Table that is passed as *transaction_head* argument is required to have following columns:

    id,
    form,
    deprel,
    feats
    
Table that is passed as *transaction_row* argument is required to have following columns:

    head_id,
    form,
    deprel,
    feats
    
Table that is passed as *patterns* argument is required to have following columns:

    pat_id,
    form,
    deprel
    


The procedure has two steps: 
first step is to find matches from *transaction_row* table. Negations that include word *ei* or a form of word *ära* can be found from this table.
Second step is to find matches from *transaction_head* table. Negated forms of verb *olema* can be found from this table, e.g. *pole*, *polnud* etc.
