## Analysis of 1000 obliques in cases denoting location

This notebook is to find how many of each annotation class is represented in 1000 obliques in location denoting cases. The classes with examples can be found at documentation/annotation_gudelines_location.md

### Example of the data

```python
from collections import Counter
import pandas as pd
```

```python
annotation_df = pd.read_csv('C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\fyysilised_kohad\\syntax_experiments_semantic_labelling\\physical_location_labelling\\results\\kohasonad_margendus.csv', sep=';')
annotation_df
```

### raw frequency

```python
annotation = annotation_df['margend'].tolist() #annotation out
sagedused = Counter(annotation)
#filtreerime highest -> lowest
filt_sagedus = sorted(sagedused.items(), key=lambda x: x[1], reverse=True)
filt_sagedus
```

### relative freequency

```python
for paar in filt_sagedus:
    print(paar[0], paar[1]/1000*100, '%')
```

### how many physical locations were placenames

```python
annotation_loc_c = annotation_df[annotation_df['margend'] == "loc_c"] 
loc_c = annotation_loc_c['form'].tolist() 
name = 0
for form in loc_c:
    if form[0].isupper():
        name += 1
print(name/len(loc_c)*100, '%')
```
