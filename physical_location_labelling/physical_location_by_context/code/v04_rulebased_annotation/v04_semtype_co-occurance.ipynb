## Counting co-occuring semantic types

This notebook finds how much each semantic type appears with location, time, state and event tags. Most common ones are gone through manually to decide if they can be combined into a single tag 

```python
import pandas as pd
```

```python
sem2 = pd.read_csv('C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\estnltk_syntax_repo_kloon\\physical_location_labelling\\physical_location_by_context\\base_data\\ekilexist\\obliikva_2_semtyyp.csv')
sem2
```

```python
def co_occurance(df, tag):
    semco = df[df[tag] == '✓']
    # Count occurrences of each class
    class_counts = semco['other_types'].value_counts()
    print(class_counts.to_string())
```

## koht/location

```python
co_occurance(sem2, 'koht')
```

## aeg/time

```python
co_occurance(sem2, 'aeg')
```

## seisund/state

```python
co_occurance(sem2, 'seisund')
```

## sündmus/event

```python
co_occurance(sem2, 'sündmus')
```
