# Lausete tekstid korpusest

See skript võtab välja koondkorpuse lausete tekstid etteantud ID-de põhjal ja salvestab nad csv faili.


```python
%%time
%load_ext autoreload
%autoreload 2
```

```python
# read ids from file 
with open('lause_id.txt', 'r') as f:
    ids = [int(line.strip()) for line in f]

print('lause id-sid', len(ids))
print('unikaalseid', len(list(set(ids))))
```

```python
from pandas import DataFrame
from estnltk.storage.postgres import PostgresStorage, IndexQuery
from datetime import datetime


collection_name = "koondkorpus_sentences"
BATCH_SIZE = 50000
date_time = datetime.now().strftime("%Y%m%d-%H%M%S")

tsv_file_name = f"laused_{date_time}.csv"

storage = PostgresStorage(
    pgpass_file="~/.pgpass",
    schema="estonian_text_corpora",
    role="estonian_text_corpora_read",
    temporary=False,
)

collection = storage[collection_name]

data = []
for collection_id, text in collection.select(progressbar="ascii", query=IndexQuery(ids)):
    data.append({'lause_id':collection_id, 'text': text.text})
len(data)

DataFrame.from_dict(data).to_csv(tsv_file_name, index=None)
print("done.")
```
