# Calculating semantic tag statistics for lemmas

This notebook does two things:
1. Calculates how much of the Estonian reference Corpus the aggregated ekilex tags cover
2. Creates a new database table with lemmas, their tag and frequency to see if the most frequent lemmas recieved an aggregated ekilex tag automatically or need further (possibly manual) tagging

```python
import sqlite3
```

```python
# database file path
filename = "../drive_data/v33_koondkorpus_transaktsioonid_v04_2.db"

# connecting with database
conn_db = sqlite3.connect(filename)
cursor_db = conn_db.cursor()
```

```python
def notnull(table, lemma_count, uniq_lemma_count, cursor):
    #repeating adverbial counts
    cursor.execute("SELECT COUNT(*) FROM " + table + " WHERE ekilex_tag IS NOT NULL")
    not_empty = cursor.fetchall()
    coverage = (not_empty[0][0]/lemma_count)*100
    print("Out of " + str(lemma_count) +" adverbials, the Ekilex tags cover " + str(not_empty[0][0]) + " aka " + str(round(coverage, 2))+ "%")

    #unique adverbial counts
    cursor.execute("SELECT COUNT(DISTINCT lemma) FROM " + table + " WHERE ekilex_tag IS NOT NULL")
    uniq_tagged_lemmas = cursor.fetchall()
    coverage_uniq = (uniq_tagged_lemmas[0][0]/uniq_lemma_count)*100
    print("Out of " + str(uniq_lemma_count) + " unique adverbials, the Ekilex tags cover " + str(uniq_tagged_lemmas[0][0]) + " aka " + str(round(coverage_uniq, 2))+ "%")
```

```python
def sempercent(table, tag, lemma_count, uniq_lemma_count, cursor):
    #repeating adverbial tag count
    cursor.execute("SELECT COUNT(*) FROM " + table + " WHERE ekilex_tag = '" + tag + "'")
    tag_count = cursor.fetchall()
    coverage = (tag_count[0][0]/lemma_count)*100
    print("Out of " + str(lemma_count) +" adverbials, the " + tag + " tag covers " + str(tag_count[0][0]) + " aka " + str(round(coverage, 2))+ "%")
    #unique adverbial tag count
    cursor.execute("SELECT COUNT(DISTINCT lemma) FROM " + table + " WHERE ekilex_tag = '" + tag + "'")
    uniq_tagged_lemmas = cursor.fetchall()
    coverage_uniq = (uniq_tagged_lemmas[0][0]/uniq_lemma_count)*100
    print("Out of " + str(uniq_lemma_count) + " unique adverbials, the " + tag + " tag covers " + str(uniq_tagged_lemmas[0][0]) + " aka " + str(round(coverage_uniq, 2))+ "%")
```

## Ekilex tag coverage

### Spatial obliques

```python
# ### How many words in spatial cases the reference corpus has
cursor_db.execute("SELECT lemma FROM spatial_obl")
lemmas = cursor_db.fetchall()
lemma_amount = len(lemmas)
uniq_lemma_amount = len(set(lemmas))
                        
notnull('spatial_obl', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('spatial_obl', 'location', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('spatial_obl', 'time', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('spatial_obl', 'event', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('spatial_obl', 'state', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('spatial_obl', 'alive', lemma_amount, uniq_lemma_amount, cursor_db)
```

```python
# ### How many words in spatial cases the reference corpus has
cursor_db.execute("SELECT lemma FROM advmod")
lemmas = cursor_db.fetchall()
lemma_amount = len(lemmas)
uniq_lemma_amount = len(set(lemmas))

#overall tag coverage
notnull('advmod', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('advmod', 'amount', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('advmod', 'general', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('advmod', 'location', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('advmod', 'manner', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('advmod', 'state', lemma_amount, uniq_lemma_amount, cursor_db)
print()
sempercent('advmod', 'time', lemma_amount, uniq_lemma_amount, cursor_db)
```

## Create new lemma+tag+lemma_frequency table in the database
Used to see if the most frequent words are covered by ekilex tags or need to be manually tagged.
Database table has unique lemmas, what ekilex tag they recieved and how many times the lemma appears in the Estonian reference corpus

```python
def newtable(file, input_table, output_table):
    conn = sqlite3.connect(file)
    cursor = conn.cursor()

    cursor.execute("DROP TABLE IF EXISTS " +  output_table)

    # Step 1: Create the new results table 
    cursor.execute(f"""
        CREATE TABLE IF NOT EXISTS {output_table} (
            lemma TEXT,
            tag TEXT,
            lemma_count INT
        )
    """)

    # Step 2: Aggregate counts and calculate percentages
    cursor.execute(f"""
        INSERT INTO {output_table} (lemma, tag, lemma_count)
        SELECT 
            lemma, 
            ekilex_tag, 
            COUNT(*) AS lemma_count
         FROM {input_table}
        GROUP BY lemma
    """)

    # Commit and close
    conn.commit()
    conn.close()
```

```python
newtable(filename, 'spatial_obl', 'lemma_frequency_obl')
```

```python
newtable(filename, 'advmod', 'lemma_frequency_adv')
```
