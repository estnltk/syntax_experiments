```python
import csv
import re
```

```python
input_file = "C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\estnltk_syntax_repo_kloon\\physical_location_labelling\\physical_locations_by_word\\classification.csv"
output_file = "C:\\Users\\kertu.saul\\OneDrive - Eesti Keele Instituut\\Dokumendid\\doktoritoo\\estnltk_syntax_repo_kloon\\physical_location_labelling\\physical_locations_by_word\\classification_clear.txt"

# read files in
classes = []
with open(input_file, mode ='r') as file:
  klassid = csv.reader(file)
  for lines in klassid:
        classes.append(lines[0])

classes
```

```python
def normalize_input(input_str):
    # Remove quotes and trailing characters after LOC or NONE
    match = re.match(r'["\']?(NONE|LOC)["\']?.*', input_str.strip(), re.IGNORECASE)
    if match:
        return match.group(1).upper()
    return None  # or raise an error / return 'UNKNOWN' if not matching

# Example inputs
normalized = [normalize_input(i) for i in classes[1:]]
normalized
```

```python
with open(output_file, 'w') as file:
    for item in normalized:
        file.write(item + '\n')  # writes each item on a new line
```
