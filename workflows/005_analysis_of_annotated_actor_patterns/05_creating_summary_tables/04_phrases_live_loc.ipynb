## Creates table for elus and koht phrases

Võtab välja need verbobl fraasid, mis on märgitud kohaks või elusaks nimekirja põhjal.

```python
import sqlite3
import pandas as pd
from tqdm import tqdm
import powerlaw as pwl
from collections import Counter
import matplotlib.pyplot as plt
import numpy as np
import sys
sys.path.append("..")
from common_display import display_db_table 
```

## Configuration

```python
DB_DIR = "../example_data"

# loodavad tabelid
PATTERN_MATCHES_DB = f"{DB_DIR}/pattern_matches.db"

LOCATION_PHRASES = "verbobl_phrases_loc"
ACTOR_PHRASES = "verbobl_phrases_live"

# table for all obl locative case verbs and roots 
TRANSACTIONS_OBL_LOC = "trans_obl_loc"
```

## Connect to database

```python
con = sqlite3.connect(PATTERN_MATCHES_DB)
cur = con.cursor()
```

## Workflow

```python
%%time

cur.execute("""DROP TABLE IF EXISTS {table}""".format(table = LOCATION_PHRASES))

cur.execute("""
Create table {new_table} as
SELECT * FROM {tbl1}
where koht = 'YES'
""".format(new_table=LOCATION_PHRASES, tbl1=TRANSACTIONS_OBL_LOC))
```

```python
display_db_table(con, LOCATION_PHRASES)
```

```python
%%time

cur.execute("""DROP TABLE IF EXISTS {table}""".format(table = ACTOR_PHRASES))

cur.execute("""
Create table {new_table} as
SELECT * FROM {tbl1}
where elus = 'YES'
""".format(new_table=ACTOR_PHRASES, tbl1=TRANSACTIONS_OBL_LOC))
```

```python
display_db_table(con, ACTOR_PHRASES)
```

```python
con.close()
```
