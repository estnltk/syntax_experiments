```python
import sqlite3
import pandas as pd
```

```python
# nimekirjade failid
elusolendid = "../source_data/elusolendid_srt_biol_elus.csv"
kohad = "../source_data/loc_lemmas_edited.csv"

# määruste andmebaas
per_loc_db = "../example_data/db_per_loc.db"

# tabelite nimed elusolendite ja kohtade jaoks
elustabel = 'elus_v1'
kohttabel = 'koht_v1'
```

```python
# uue andmebaasi loomine
con = sqlite3.connect(per_loc_db)
cur = con.cursor()
cur.execute('pragma encoding=UTF8')
```

```python
elus = pd.read_csv(elusolendid, sep=",", encoding="utf-8")
#elus
elus2 = elus.drop(["cnt"], axis=1)
#elus2
```

```python
elus2.to_sql(name=elustabel, con=con)
con.commit()
```

```python
koht = pd.read_csv(kohad, sep=",", encoding="utf-8", header=None)
koht = koht.rename(columns={0: "lemma"})
#kohad
```

```python
koht.to_sql(name=kohttabel, con=con)
con.commit()
```

```python
con.close()
```
