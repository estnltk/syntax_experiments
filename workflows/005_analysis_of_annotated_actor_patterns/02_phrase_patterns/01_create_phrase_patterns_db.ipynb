## Creates two tables from lists that can be used to annotate root words as koht/elus

### these tables are used to create enriched_transactions table

```python
import sqlite3
import pandas as pd
```

## Configuration

```python
DB_DIR = "../example_data"
SOURCE_DIR = "../source_data"

# nimekirjade failid
ELUSOLENDID = f"{SOURCE_DIR}/elusolendid_srt_biol_elus.csv"
KOHAD = f"{SOURCE_DIR}/loc_lemmas_edited.csv"

# määruste andmebaas
PHRASE_PATTERNS = f"{DB_DIR}/phrase_patterns.db"

# tabelite nimed elusolendite ja kohtade jaoks
ELUSTABEL = 'elus_v1'
KOHTTABEL = 'koht_v1'
```

## Connect to db

```python
con = sqlite3.connect(PHRASE_PATTERNS)
cur = con.cursor()
```

## Workflow

### Read in lists

```python
elus = pd.read_csv(ELUSOLENDID, sep=",", encoding="utf-8")
#elus
elus2 = elus.drop(["cnt"], axis=1)
#elus2
```

```python
koht = pd.read_csv(KOHAD, sep=",", encoding="utf-8", header=None)
koht = koht.rename(columns={0: "lemma"})
#kohad
```

### Save lists to db

```python
elus2.to_sql(name=ELUSTABEL, con=con)
con.commit()
```

```python
koht.to_sql(name=KOHTTABEL, con=con)
con.commit()
```

```python
con.close()
```
