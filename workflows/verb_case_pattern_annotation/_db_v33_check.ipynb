# Checking tables that are created in prep and during annotation

## example, tested on another db file, some results might differ from other notebooks

```python
import sqlite3
import pandas as pd
import configparser
```

```python
def load_config(path):
    """Loads config.
    """
    config = configparser.ConfigParser()
    status = config.read(path) 
    assert status == [path]
    return config
```

```python
config_f = "../conf/conf_ex_A.ini"
config = load_config(config_f)

# database file path
DB_FILE = config["configuration"]["database"]
#print(DB_FILE)
```

```python
# connecting with database
conn = sqlite3.connect(DB_FILE)
cur = conn.cursor()
```

```python
q1 = """
SELECT * FROM sqlite_master WHERE type='table'
"""
tables = pd.read_sql(q1, conn)
tables
```

```python
### transaction_row

query = f"SELECT count(*) FROM transaction_row"
tr_rows = pd.read_sql(query, conn)
print(tr_rows)

query = f"SELECT * FROM transaction_row limit 10"
tr = pd.read_sql(query, conn)
tr
```

```python
### transaction_head

query = f"SELECT count(*) FROM transaction_head"
th_rows = pd.read_sql(query, conn)
print(th_rows)

query = f"SELECT * FROM transaction_head limit 10"
th = pd.read_sql(query, conn)
th
```

```python
### spatial_obl

query = f"SELECT count(*) FROM spatial_obl"
so_rows = pd.read_sql(query, conn)
print(so_rows)

query = f"SELECT * FROM spatial_obl limit 10"
so = pd.read_sql(query, conn)
so
```
