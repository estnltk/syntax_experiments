```python
import sqlite3
import pandas as pd
import os
import csv
```

```python

DATABASE = "../../drive_data_v1/v33_koondkorpus_sentences_verb_pattern_obl_20241002-130310_test.db"

OBL_TABLE = "spatial_obl"

FILTERED_CLASS_TABLE = f"spatial_obl_n_random_100k_lemma"

FNAME = f"n_random_100k_lemma"
LARGE_DATA_FILE = f"../data/n_random/{FNAME}.csv"
LARGE_DATA_FILE_SORTED = f"../data/n_random/{FNAME}_sorted.csv"
```

### andmetabelid

```python
conn = sqlite3.connect(DATABASE)
cursor = conn.cursor()
```

### 100K random lemmat ja nende laused

```python

query = f"""SELECT *FROM {OBL_TABLE}"""

filtered_class = pd.read_sql(query, conn)
filtered_class
```

```python
# shuffle
#df = filtered_class.groupby("lemma").sample(frac=1, random_state=1, replace=False)
sample = (
    filtered_class.sample(frac=1, random_state=1)
      .drop_duplicates("lemma")
      .head(100000)
)
```

```python
sample.to_sql(FILTERED_CLASS_TABLE, conn, if_exists="replace", index=False)
```

```python
sample.to_csv(LARGE_DATA_FILE, encoding="utf-8", index = False,sep=",", quoting=csv.QUOTE_MINIMAL)
```
