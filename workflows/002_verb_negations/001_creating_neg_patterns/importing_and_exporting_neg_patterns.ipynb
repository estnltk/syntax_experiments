## Importing and exporting negation patterns

Importing CSV-file with negation patterns from *source_data* and saving it as a table to a new database. This new database will be used in next processing steps.

```python
import sys
sys.path.append('../../../common_code')
```

```python
import sqlite3
import pandas as pd
from db_operations.db_display import *
```

## Input parameters

```python
INPUT_DIR = "../source_data"
RESULT_DIR = "../example_data"

INPUT_FILE = f"{INPUT_DIR}/neg_patterns.csv"
RESULT_DB = f"{RESULT_DIR}/neg_patterns.db"
```

## Data processing

```python
con = sqlite3.connect(RESULT_DB)
cur = con.cursor()

df = pd.read_csv(INPUT_FILE, encoding='UTF-8')

df.to_sql('neg_patterns', con, index=False)

con.close()
```
