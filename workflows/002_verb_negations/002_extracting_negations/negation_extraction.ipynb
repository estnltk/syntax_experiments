### Extracting negations

Extracting indexes of transactions that contain a match to existing negation patterns in neg_tables. Function `extract_negations()` extracts ID-s of transactions that contain a verb negation based on given negation patterns. The result is a new table of ID-s of transaction matches and corresponding pattern ID-s.

    
The procedure has two steps: 
first step is to find matches from *transaction_row* table. Negations that include word *ei* or a form of word *ära* can be found from this table.
Second step is to find matches from *transaction_head* table. Negated forms of verb *olema* can be found from this table, e.g. *pole*, *polnud* etc.
