## Extracting negations

Extracting indexes of transactions that contain a match to existing negation patterns in neg_tables. The result is a new table of ID-s of transaction matches and corresponding pattern ID-s.

    
The procedure has two steps: 
first step is to find matches from *transaction_row* table. Negations that include word *ei* or a form of word *ära* can be found from this table.
Second step is to find matches from *transaction_head* table. Negated forms of verb *olema* can be found from this table, e.g. *pole*, *polnud* etc.

## Input parameters

```python
INPUT_DIR = "C:/Users/liivas/Documents/Töö/verbirektisoonid"

RESULT_DB = "negations.db"
NEG_PATTERNS_DB = f"{INPUT_DIR}/db_operations/workflows/002_verb_negations/001_creating_neg_patterns/neg_patterns.db"
TRANSACTION_DB = f"{INPUT_DIR}/v32_data.db"
```

## Data processing

```python
con = sqlite3.connect(RESULT_DB)
cur = con.cursor()

cur.execute(f'ATTACH DATABASE "{NEG_PATTERNS_DB}" AS neg_patterns')
cur.execute(f'ATTACH DATABASE "{TRANSACTION_DB}" AS transactions')

extract_negations(
    conn=con,
    cur=cur,
    transaction_head='transactions.transaction_head',
    transaction_row='transactions.transaction_row',
    patterns='neg_patterns.neg_patterns',
    output_table='neg_phrase_matches',
    overwrite=True
    )
con.close()
```
