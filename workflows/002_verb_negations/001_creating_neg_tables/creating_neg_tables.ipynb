```python
# **Input parameters**

result_db = "neg_tables.db"
vp_data3 = "C:/Users/liivas/Documents/Töö/verbirektisoonid/vp_data3.db"
v32_data = "C:/Users/liivas/Documents/Töö/verbirektisoonid/v32_data.db"

# **Data processing**

#creating new database where resulting tables will be stored
con = sqlite3.connect(result_db)
cur = con.cursor()

#attaching verb patterns database
cur.execute(f'ATTACH DATABASE "{vp_data3}" AS vp')

#attaching transactions database (currently v32)
cur.execute(f'ATTACH DATABASE "{v32_data}" AS v32')

# **Documentation**

?create_neg_tables

# **Implementation**

create_neg_tables(
    cur,
    verb_matches='vp.verb_matches',
    transaction_head='v32.transaction_head',
    transaction_row='v32.transaction_row'
    )
con.close()
```
