## Creating negation tables

Creating a new database with three verb negation tables. The tables will be created from transactions (v32) that match a pattern in verb patterns database (vp_data3).

```python
import sqlite3
from paths import PATH_ROOT
from db_operations.verb_transactions.neg_tables_creation import *
from db_operations.db_display import *
```

## Input parameters

```python
INPUT_DIR = "C:/Users/liivas/Documents/Töö/verbirektisoonid"

RESULT_DB = "neg_tables.db"
PATTERNS_DB = f"{INPUT_DIR}/vp_data3.db"
TRANSACTION_DB = f"{INPUT_DIR}/v32_data.db"
```

## Data processing

```python
con = sqlite3.connect(RESULT_DB)
cur = con.cursor()

cur.execute(f'ATTACH DATABASE "{PATTERNS_DB}" AS patterns')
cur.execute(f'ATTACH DATABASE "{TRANSACTION_DB}" AS transactions')

create_neg_tables(
    con,
    cur,
    verb_matches='patterns.verb_matches',
    transaction_head='transactions.transaction_head',
    transaction_row='transactions.transaction_row'
    )
con.close()
```
