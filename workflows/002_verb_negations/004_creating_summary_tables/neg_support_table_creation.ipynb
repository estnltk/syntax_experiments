## Creating negation support table

Creating a new summary table based on *negation_matches.db* that shows both total verb / verb compound frequencies and negated form frequencies among verbs / verb compounds in transactions and also relative negation support for each verb / verb compound.

```python
import sys

sys.path.append('../../../common_code')
```

```python
import sqlite3
from paths import PATH_ROOT
from db_operations.verb_transactions.neg_support_creation import create_verb_neg_support_table
from db_operations.db_display import *
```

## Input parameters

```python
INPUT_DIR = "../example_data"

RESULT_DB = f"{INPUT_DIR}/neg_support.db"
NEG_MATCHES_DB = f"{INPUT_DIR}/negation_matches.db"
TRANSACTION_DB = f"{INPUT_DIR}/transactions.db"
```

## Data processing

```python
con = sqlite3.connect(RESULT_DB)
cur = con.cursor()

cur.execute(f'ATTACH DATABASE "{NEG_MATCHES_DB}" AS neg_matches')
cur.execute(f'ATTACH DATABASE "{TRANSACTION_DB}" AS transactions')

create_verb_neg_support_table(
    cur,
    transaction_head='transactions.transaction_head',
    verb_neg_matches='neg_matches.neg_phrase_matches',
    output_table='verb_neg_support'
)
```
