### II Extracting elative construction pattern matches

```python
# pattern of elative construction
el_conx_pattern = {
        "nodes": {
            "A": {
                "deprels": ["nmod"], # A has to be a substantive and a nominal modifier in singular elative case
                "partofspeeches": ["S"],
                "forms": ["sg el"],
                "lemmas": None,
                "lemma_regex": None,
                "instances": None
                },
            "B": {
                "deprels": None,
                "partofspeeches": ["S"], # B has to be a substantive
                "forms": None,
                "lemmas": None,
                "lemma_regex": None,
                "instances": None
                }
        },
        "node_restrictions": {
            "A": None,
            "B": {"deprels": ["nmod"], # B can't be a nominal modifier, in terminative case, nor can it's lemma end with "mine"-clitic
                  "partofspeeches": None,
                  "forms": ["sg ter", "pl ter"],
                  "lemmas": None,
                  "lemma_regex": [r".*mine$"],
                  "instances": None,
                  "parent": None,
                  "child": None
                }
        },
        "edges": [
            ("B", "A")  # B is head of A
        ],
        "order": [
            ("A", "BEFORE", "B") # A has to appear before B in text
        ]
    }
```

```python
# finding elative construction instances from sentences based on pattern
match_sentences = []
matches = []

for idx, sentence in enumerate(data):

    sent_matches = get_pattern(stanza_words = sentence.v172_stanza_syntax, 
                                        pattern = el_conx_pattern, 
                                        anchor_name = "B")
    if len(sent_matches) > 0:
        match_sentences.append(sentence.v172_stanza_syntax)
        matches.append(sent_matches)
```

```python
# list of specified members of construction examples
conx_member_list = [[get_conx_member(match, "A") for idx2, match in enumerate(matches[idx1])] for idx1, sent in enumerate(match_sentences)]
print(conx_member_list[0])

# list of specified members of construction examples as lemmas
conx_member_lemma_list = [[member.morph_analysis.lemma[0] for member in members] for idx1, members in enumerate(conx_member_list)]
print(conx_member_lemma_list[0])

# flattened version of member lemma list
lemma_list = [member for members in conx_member_lemma_list for member in members]
print(lemma_list[0])
```

```python
# Lists of material and occupation/nationality/role lemmas
SEM_DATA_PATH = "../data/semantic_categories/"
FILENAME1 = "material.txt"
FILENAME2 = "occupation_nationality_role.txt"

material: list[str] = read_txt_to_list(f"{SEM_DATA_PATH}{FILENAME1}")
occupation_nationality_role: list[str] = read_txt_to_list(f"{SEM_DATA_PATH}{FILENAME2}")
```

```python
# Filtering and categorizing pattern match lemmas
elative_material_attribute: list[str] = get_filtered_list(lemma_list, material+occupation_nationality_role)
```

```python
# pattern match lemmas that do not occur among material nor attribute lemmas
elative_other = list(set(lemma_list).difference(set(elative_material_attribute)))
```

```python
phrases = []
for idx1, sent_matches in enumerate(matches):
    for idx2, match in enumerate(sent_matches):
        if match["A"].lemma in elative_material_attribute:
            phrases.append("##".join([match["A"].text, match["B"].text]))
```

```python
phrases2 = []
for idx1, sent_matches in enumerate(matches):
    for idx2, match in enumerate(sent_matches):
        if match["A"].lemma in elative_other:
            phrases2.append("##".join([match["A"].text, match["B"].text]))
```

```python
data = []

for idx, phrase_freq in enumerate(phrase_freqs):
    phrase = phrase_freq[0].split("##")   
    if len(phrase) != 2:
        continue   
    freq = phrase_freq[1]
    data_dct = {
                "member1": phrase[0], 
                "member2": phrase[1],
                "freq": freq
                }
    data.append(data_dct)
```

```python
data2 = []

for idx, phrase_freq in enumerate(phrase_freqs2):
    phrase = phrase_freq[0].split("##")   
    if len(phrase) != 2:
        continue   
    freq = phrase_freq[1]
    data_dct = {
                "member1": phrase[0], 
                "member2": phrase[1],
                "freq": freq
                }
    data2.append(data_dct)
```
