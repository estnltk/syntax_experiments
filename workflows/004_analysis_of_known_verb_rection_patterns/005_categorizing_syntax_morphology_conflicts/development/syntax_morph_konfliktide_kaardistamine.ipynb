### Defineeritud vigade eraldamine-arvutamine

```python
con = sqlite3.connect(f"{SOURCE_DATA_PATH}{CONFLICTS}")
cur = con.cursor()

cur.execute(f'ATTACH DATABASE "{SOURCE_DATA_PATH}{SENTENCES}" AS sents')
cur.execute(f'ATTACH DATABASE "{SOURCE_DATA_PATH}{RESULTS}" AS result')

cur.execute("""
DROP TABLE IF EXISTS result.nsubj_not_nom_part
""")

cur.execute("""
CREATE TABLE result.nsubj_not_nom_part AS
SELECT
    tbl1.id,
    tbl1.pattern_id,
    tbl1.sentence_id,
    tbl1.verb_loc,
    tbl1.compound_loc,
    tbl1.phrase_root_loc,
    tbl1.verb_phrase_loc,
    tbl1.phrase_case,
    tbl1.phrase_deprel,
    tbl1.verb,
    tbl1.verb_compound,
    tbl1.phrase,
    tbl1.phrase_root_lemma,
    tbl1.current_analysis,
    tbl1.current_case,
    tbl1.possible_cases,
    tbl2.sentence
FROM
(
    SELECT * FROM
        syntax_morphology_conflicts
    WHERE
        phrase_deprel = "nsubj"
    AND
        current_case != "nom"
    AND
        current_case != "part"
) AS tbl1
INNER JOIN
(
    SELECT
        id,
        text AS sentence
    FROM
        sents.sentences
) AS tbl2
ON
tbl1.sentence_id=tbl2.id 
""")

con.close()
```

```python
# obj ^(nom|gen|part)
con = sqlite3.connect(f"{SOURCE_DATA_PATH}{CONFLICTS}")
cur = con.cursor()

cur.execute(f'ATTACH DATABASE "{SOURCE_DATA_PATH}{SENTENCES}" AS sents')
cur.execute(f'ATTACH DATABASE "{SOURCE_DATA_PATH}{RESULTS}" AS result')

cur.execute("""
DROP TABLE IF EXISTS result.obj_not_nom_gen_part
""")

cur.execute("""
CREATE TABLE result.obj_not_nom_gen_part AS
SELECT
    tbl1.id,
    tbl1.pattern_id,
    tbl1.sentence_id,
    tbl1.verb_loc,
    tbl1.compound_loc,
    tbl1.phrase_root_loc,
    tbl1.verb_phrase_loc,
    tbl1.phrase_case,
    tbl1.phrase_deprel,
    tbl1.verb,
    tbl1.verb_compound,
    tbl1.phrase,
    tbl1.phrase_root_lemma,
    tbl1.current_analysis,
    tbl1.current_case,
    tbl1.possible_cases,
    tbl2.sentence
FROM
(
    SELECT * FROM
        syntax_morphology_conflicts
    WHERE
        phrase_deprel = "obj"
    AND
        current_case != "nom"
    AND
        current_case != "gen"
    AND
        current_case != "part"
) as tbl1
INNER JOIN
(
    SELECT
        id,
        text AS sentence
    FROM
        sents.sentences
) AS tbl2
ON
tbl1.sentence_id=tbl2.id   
""")

con.close()
```

```python
# advcl gen
con = sqlite3.connect(f"{SOURCE_DATA_PATH}{CONFLICTS}")
cur = con.cursor()

cur.execute(f'ATTACH DATABASE "{SOURCE_DATA_PATH}{SENTENCES}" AS sents')
cur.execute(f'ATTACH DATABASE "{SOURCE_DATA_PATH}{RESULTS}" AS result')

cur.execute("""
DROP TABLE IF EXISTS result.advcl_gen
""")

cur.execute("""
CREATE TABLE result.advcl_gen AS
SELECT
    tbl1.id,
    tbl1.pattern_id,
    tbl1.sentence_id,
    tbl1.verb_loc,
    tbl1.compound_loc,
    tbl1.phrase_root_loc,
    tbl1.verb_phrase_loc,
    tbl1.phrase_case,
    tbl1.phrase_deprel,
    tbl1.verb,
    tbl1.verb_compound,
    tbl1.phrase,
    tbl1.phrase_root_lemma,
    tbl1.current_analysis,
    tbl1.current_case,
    tbl1.possible_cases,
    tbl2.sentence
FROM
(
    SELECT * FROM          
    syntax_morphology_conflicts
    WHERE
    phrase_deprel = "advcl"
    AND
    current_case = "gen"
) as tbl1
INNER JOIN
(
    SELECT
        id,
        text AS sentence
    FROM
        sents.sentences
) AS tbl2
ON
tbl1.sentence_id=tbl2.id
""")

con.close()
```
