## Corrections

In initially created verb pattern tables, the compound parts of longer verb constructions were separated into different columns.
Since in transactions database the compound parts are saved into the same column (as a singular string with compound parts separated by ', ' and ordered alphabetically), the representation in verb pattern tables will also be corrected by joining compound part columns.

```python
import pandas as pd
```

## Input parameters

```python
INPUT_DIR = "../source_data/rection_patterns_in_standard_format"

INPUT_FILE_LEN1 = f"{INPUT_DIR}/verb_patterns_len1_new.csv"
INPUT_FILE_LEN2 = f"{INPUT_DIR}/verb_patterns_len2_new.csv"
RESULT_FILE_LEN1 = f"{INPUT_DIR}/verb_patterns_len1_new_correct.csv"
RESULT_FILE_LEN2 = f"{INPUT_DIR}/verb_patterns_len2_new_correct.csv"
```

## Data processing

```python
df_len1 = pd.read_csv(INPUT_FILE_LEN1)
df_len2 = pd.read_csv(INPUT_FILE_LEN2)

df_len1 = df_len1.fillna('')
df_len2 = df_len2.fillna('')

def join_and_sort_compound_prt(compound_prt_list):
    compound_prt_list = [el for el in compound_prt_list if el!='']
    sorted_lst = sorted(compound_prt_list)
    return ', '.join(sorted_lst)

def get_sorted_strings(df):
    sorted_strs = []
    for idx, row in df.iterrows():
        sorted_strs.append(join_and_sort_compound_prt([row['compound_prt1'], row['compound_prt2'], row['compound_prt3']]))       
    return sorted_strs

sorted_strs_len1 = get_sorted_strings(df_len1)
sorted_strs_len2 = get_sorted_strings(df_len2)

df_len1.insert(4, 'verb_compound', sorted_strs_len1)
df_len2.insert(4, 'verb_compound', sorted_strs_len2)

df_len1['verb_compound'] = sorted_strs_len1
df_len2['verb_compound'] = sorted_strs_len2

df_len1 = df_len1.drop(['compound_prt1', 'compound_prt2', 'compound_prt3'], axis=1)
df_len2 = df_len2.drop(['compound_prt1', 'compound_prt2', 'compound_prt3'], axis=1)

df_len1 = df_len1.drop('Unnamed: 0', axis=1)
df_len2 = df_len2.drop('Unnamed: 0', axis=1)

df_len1.to_csv(RESULT_FILE_LEN1, encoding='utf-8')
df_len2.to_csv(RESULT_FILE_LEN2, encoding='utf-8')
```
