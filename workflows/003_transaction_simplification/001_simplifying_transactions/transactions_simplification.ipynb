## Simplifying transactions

Simplifying existing transactions by removing rows that contain unwanted syntactic relations (*deprel* values). The modifications will be conducted on *transaction_row* table. Currently, *deprel* values *nsubj*, *advmod*, *advcl* and *csubj* will be removed. In addition, auxiliary verbs will also be removed. 

Examples:

    "Solist [nsubj] ei osanud laulda" -> "ei osanud laulda"
    "Ebastabiilsus on [aux] õpetanud elama" -> "Ebastabiilsus õpetanud elama"

A copy of the same transactions subset database will be used as an example that was used in `002_verb_negations` workflow.

Modified transactions can be used to get a cleaner result when implementing apriori (unwanted *deprel* values will not occur in output).

## Input parameters

```python
INPUT_DIR = "../example_data"

TRANSACTION_DB = f"{INPUT_DIR}/transactions.db"
```

## Data processing

```python
con = sqlite3.connect(TRANSACTION_DB)
cur = con.cursor()

remove_deprel_from_transaction_row(cur,
                                   transaction_row='transaction_row',
                                   deprel='nsubj')
remove_deprel_from_transaction_row(cur,
                                   transaction_row='transaction_row',
                                   deprel='advmod')
remove_deprel_from_transaction_row(cur,
                                   transaction_row='transaction_row',
                                   deprel='advcl')
remove_deprel_from_transaction_row(cur,
                                   transaction_row='transaction_row',
                                   deprel='csubj')
remove_aux_verbs(cur,
                 transaction_row='transaction_row')

con.close()
```
