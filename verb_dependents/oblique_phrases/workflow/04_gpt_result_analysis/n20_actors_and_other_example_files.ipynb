```python
import pandas as pd
from tqdm import tqdm
import os
import json
import csv
import sys
```

```python
pd.set_option('display.max_colwidth', None)
pd.set_option("display.show_dimensions", True)
```

```python
RESULTS_DIR = "../../results/"

GPT_ANSWER_FILE = RESULTS_DIR + "n20_examples_large_v01/gpt_v02/"+ "gpt_b10_run01.csv"

RESULTS_DIR2 = RESULTS_DIR + "n20_examples_large_v01/result_case_samples/"
```

```python
df1 = pd.read_csv(GPT_ANSWER_FILE, encoding="utf-8",  sep=",")
```

```python
actors = df1[df1["classification2"]=="actor"]
```

```python
grouped_dfs = {
    value: group.head(100)
    for value, group in actors.groupby("morph_case")
}
```

```python
grouped_dfs.keys()
```

```python
abl = grouped_dfs["abl"]
ad = grouped_dfs["ad"]
adit = grouped_dfs["adit"]
all_ = grouped_dfs["all"]
el = grouped_dfs["el"]
ill = grouped_dfs["ill"]
in_ = grouped_dfs["in"]
```

```python
abl.to_csv(RESULTS_DIR2 + "actors_abl.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
ad.to_csv(RESULTS_DIR2 + "actors_ad.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
adit.to_csv(RESULTS_DIR2 + "actors_adit.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
all_.to_csv(RESULTS_DIR2 + "actors_all.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
el.to_csv(RESULTS_DIR2 + "actors_el.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
ill.to_csv(RESULTS_DIR2 + "actors_ill.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
in_.to_csv(RESULTS_DIR2 + "actors_in.csv", encoding="utf-8", index=False, sep=",", quoting=csv.QUOTE_MINIMAL)
```
