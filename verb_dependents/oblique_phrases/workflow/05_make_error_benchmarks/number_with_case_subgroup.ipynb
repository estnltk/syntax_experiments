```python
RESULT_DIR = "../results/"

DATA_FILE = RESULT_DIR+ "n80_examples_large_v01/gpt_v02/" + "gpt_10K_b10_run01.csv"

BM_DIR = "../locative_adverbial/benchmarks/syntax_errors/numbers_with_case/"
NUMBERS_FILE = BM_DIR + "gpt_10k_b10_run01_no_numbers_w_case.csv"
```

```python
df1 = pd.read_csv(DATA_FILE, encoding="utf-8", sep=",")
df = df1[df1["classification2"]=="no"]
```

```python
new_df = pd.DataFrame(columns = list(df.columns))

for i in range(len(df)):
    form = df.iloc[i]["form"]
    has_number = any(char.isdigit() for char in form)
    if has_number:
        new_df = pd.concat([new_df, df.iloc[[i]] ], axis=0, ignore_index=True)
```
