```python
from estnltk import Text
from estnltk.taggers import VabamorfTagger, VabamorfAnalyzer
from estnltk_neural.taggers import StanzaSyntaxTagger
from estnltk.converters import text_to_json, json_to_text
import sys, os
import re
import csv
import pandas as pd
import json
from tqdm import tqdm
import estnltk
```

```python
pd.set_option('display.max_colwidth', None)
pd.set_option("display.show_dimensions", True)
```

```python
BM_DIR = "../../locative_adverbial/benchmarks/morph_syntax_errors/"

YHILDUMINE_DIR = BM_DIR + "yhildumine/" 
YHILDUMINE = YHILDUMINE_DIR + "gpt_10K_b10_run01_no_yhildumine.csv"

VKESKSONA_DIR = BM_DIR + "v_kesksona/" 
VKESKSONA = VKESKSONA_DIR + "gpt_10K_b10_run01_vkesksona.csv"

NUM_DIR = BM_DIR + "numbers_with_case/" 
NUMBERS = NUM_DIR + "gpt_10k_b10_run01_no_numbers_w_case.csv"

MORPH_DIR = BM_DIR + "morphology_errors/" 
MORPH = MORPH_DIR + "gpt_10k_b10_run01_morph_syntax_errors.csv"
```

```python
morf_tagger = VabamorfAnalyzer(output_layer='morph_analysis')
stanza_tagger = StanzaSyntaxTagger(input_type='morph_extended', input_morph_layer='morph_extended')
```

## Ühildumine

```python
df = pd.read_csv(YHILDUMINE, sep=",", encoding="utf-8")
```

```python
texts_list = []

for i in range(len(df)):
    ex = df.iloc[[i]]
    
    txt = Text(ex.iloc[0]["sentence"])
    txt.tag_layer("words")
    txt.tag_layer("sentences")
    morf_tagger.tag(txt)
    txt.tag_layer('morph_extended')
    stanza_tagger.tag( txt )
    
    meta = {"sentence_id":int(ex.iloc[0]["sentence_id"]) , "head_id":int(ex.iloc[0]["head_id"]) , 
            "head_loc": int(ex.iloc[0]["head_loc"]), "verb": ex.iloc[0]["verb"], 
            "verb_compound": ex.iloc[0]["verb_compound"], "morph_case": ex.iloc[0]["morph_case"],
           "form": ex.iloc[0]["form"], "lemma": ex.iloc[0]["lemma"]}
    
    txt.meta = meta
    
    texts_list.append(txt)
```

```python
for i, txt in enumerate(texts_list):
    text_to_json(txt, file=f"{YHILDUMINE_DIR}/text_{i}.json")
```
